```python
# bind can devices
# get the list of all can devices using ip l
import subprocess
def get_can_devices():
    result = subprocess.check_output(['ip', 'link']).decode('utf-8')
    devices = []
    for line in result.split('\n'):
        if 'can' in line:
            device = line.split(':')[1].strip()
            devices.append(device)
    return devices
can_devices = get_can_devices()
print(f"CAN devices: {can_devices}")
# bind can devices
```
